```python
import time
import numpy as np
import xarray as xr
from gsw import f
```

```python
g = 9.8
rho0 = 1035.0
rho1 = 1036.02
rho2 = 1036.88
delta_rho = rho2 - rho1
omega = 2 * np.pi / (12.4206014*3600)
f_loc = f(45)
print(f_loc)
```

1675m is the depth of the interface

```python
lambda_theory = 2 * np.pi * g * delta_rho / rho0 * (5500 - 1675) / np.sqrt(omega**2-f_loc**2)
print('%.2f km'%(lambda_theory/1e+3))
```

```python
eta_prime = 5
eta_surface = eta_prime * delta_rho / rho0 * 1000 # convert units from m to mm
print('%.2f mm'%eta_surface)
```
